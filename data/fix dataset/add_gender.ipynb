```python
import pandas as pd
import os
from datetime import datetime

# Define base paths
base_dir = r"C:\Users\anouk\OneDrive - UvA\Documenten\GitHub\tweede-kamer-real\tweede-kamer\tweede-kamer"

# Define paths to input files
career_path = os.path.join(base_dir, "party affiliations", "combined_members_2006-present.csv")
speeches_path = os.path.join(base_dir, "data", "speeches2014-2024", 
                           "party_affilliations_fixed", 
                           "speeches_2014-2024_speakers_parties_fixed_with_ids.csv")

# Load datasets
careers_df = pd.read_csv(career_path, sep=';')
speeches_df = pd.read_csv(speeches_path, sep=';')

# Verify loading
print("Careers dataset shape:", careers_df.shape)
print("Speeches dataset shape:", speeches_df.shape)

# Display first few rows of each
print("\nCareers dataset preview:")
print(careers_df.head())
print("\nSpeeches dataset preview:")
print(speeches_df.head())
```

```python
# Clean and normalize names for matching
def clean_name(name):
    """Remove titles and normalize names"""
    if pd.isna(name) or name.lower() == 'de voorzitter':
        return None
        
    # Remove common titles
    titles = ['Minister', 'Staatssecretaris', 'De heer', 'Mevrouw', 'President']
    name = name.strip()
    for title in titles:
        name = name.replace(title, '').strip()
    
    return name.strip()

# Create career timeline and verify party matches
def create_career_timeline(careers_df):
    """Create timeline of political careers with party verification"""
    # Convert dates to datetime
    careers_df['start_date'] = pd.to_datetime(careers_df['start_date'], format='%d %B %Y')
    careers_df['end_date'] = pd.to_datetime(careers_df['end_date'], format='%d %B %Y')
    
    # Clean names and create mapping of name to parties
    careers_df['clean_name'] = careers_df['full_name'].apply(clean_name)
    
    # Create dict of valid parties for each name
    name_party_dict = {}
    for name, group in careers_df.groupby('clean_name'):
        name_party_dict[name] = set(group['party'].unique())
    
    return careers_df, name_party_dict

# Calculate seniority for each speech
def calculate_seniority(speeches_df, careers_df, name_party_dict):
    """Calculate seniority while verifying party matches"""
    # Clean names in speeches dataset
    speeches_df['clean_name'] = speeches_df['speaker_name'].apply(clean_name)
    
    # Convert speech date to datetime
    speeches_df['date'] = pd.to_datetime(speeches_df['date'])
    
    # Initialize seniority column
    speeches_df['seniority_days'] = None
    
    # Track skipped speakers for logging
    skipped_speakers = set()
    
    for idx, row in speeches_df.iterrows():
        if row['clean_name'] is None:
            continue
            
        # Check if speaker exists in career data
        if row['clean_name'] not in name_party_dict:
            skipped_speakers.add(f"{row['speaker_name']} - Not found in career data")
            continue
            
        # Verify party match
        speaker_career = careers_df[careers_df['clean_name'] == row['clean_name']]
        valid_parties = name_party_dict[row['clean_name']]
        
        if pd.notna(row['speaker_party']) and row['speaker_party'] not in valid_parties:
            skipped_speakers.add(f"{row['speaker_name']} - Party mismatch: {row['speaker_party']} vs {valid_parties}")
            continue
        
        # Get first start date considering 2-year break rule
        speaker_dates = speaker_career.sort_values('start_date')
        current_start = speaker_dates['start_date'].iloc[0]
        
        # Calculate seniority
        seniority = (row['date'] - current_start).days
        speeches_df.at[idx, 'seniority_days'] = max(0, seniority)
    
    # Print skipped speakers report
    print("\nSkipped speakers report:")
    for speaker in sorted(skipped_speakers):
        print(f"- {speaker}")
        
    return speeches_df
```

```python
# Add this function to handle Dutch dates
def convert_dutch_date(date_str):
    """Convert Dutch date string to datetime"""
    dutch_months = {
        'januari': 'January',
        'februari': 'February',
        'maart': 'March',
        'april': 'April',
        'mei': 'May',
        'juni': 'June',
        'juli': 'July',
        'augustus': 'August',
        'september': 'September',
        'oktober': 'October',
        'november': 'November',
        'december': 'December'
    }
    
    if pd.isna(date_str):
        return pd.NaT
    
    # Split the date string
    parts = date_str.split()
    if len(parts) != 3:
        return pd.NaT
        
    day, month, year = parts
    
    # Convert Dutch month to English
    month = dutch_months.get(month.lower(), month)
    
    # Reconstruct date string in English
    english_date = f"{day} {month} {year}"
    
    return pd.to_datetime(english_date, format='%d %B %Y')

# Modify the create_career_timeline function
def create_career_timeline(careers_df):
    """Create timeline of political careers with party verification"""
    # Convert dates using the Dutch date converter
    careers_df['start_date'] = careers_df['start_date'].apply(convert_dutch_date)
    careers_df['end_date'] = careers_df['end_date'].apply(convert_dutch_date)
    
    # Clean names and create mapping of name to parties
    careers_df['clean_name'] = careers_df['full_name'].apply(clean_name)
    
    # Create dict of valid parties for each name
    name_party_dict = {}
    for name, group in careers_df.groupby('clean_name'):
        name_party_dict[name] = set(group['party'].unique())
    
    return careers_df, name_party_dict
```

```python
# Run the seniority calculation pipeline
# Process career data
careers_df, name_party_dict = create_career_timeline(careers_df)

# Calculate seniority
speeches_df = calculate_seniority(speeches_df, careers_df, name_party_dict)

# Fill missing values with -1 to indicate no seniority data
speeches_df['seniority_days'] = speeches_df['seniority_days'].fillna(-1)

# Display some statistics
print("\nSeniority statistics:")
print(speeches_df[speeches_df['seniority_days'] >= 0]['seniority_days'].describe())

# Display sample of results
print("\nSample of speeches with seniority:")
sample_cols = ['speaker_name', 'speaker_party', 'date', 'seniority_days']
print(speeches_df[speeches_df['seniority_days'] >= 0][sample_cols].head(10))
```

```python
def clean_name_for_matching(name):
    """Clean and standardize names for matching"""
    if pd.isna(name) or name.lower() == 'de voorzitter':
        return None
        
    # Convert to lowercase for consistent matching
    name = name.lower().strip()
    
    # Remove standard prefixes
    prefixes = ['minister', 'staatssecretaris', 'de heer', 'mevrouw', 'president', 'kamerlid']
    for prefix in prefixes:
        if name.startswith(prefix):
            name = name[len(prefix):].strip()
    
    # Handle special cases with 'van', 'van der', etc.
    if ' van ' in name or ' van der ' in name or ' de ' in name:
        return name  # Keep full name for these cases
    
    # Get last name for simple cases
    parts = name.split()
    return parts[-1].strip()

def create_career_timeline(careers_df):
    """Create timeline of political careers with party verification"""
    # Convert dates using the Dutch date converter
    careers_df['start_date'] = careers_df['start_date'].apply(convert_dutch_date)
    careers_df['end_date'] = careers_df['end_date'].apply(convert_dutch_date)
    
    # Clean names and create mapping including party info
    careers_df['clean_name'] = careers_df['full_name'].apply(clean_name_for_matching)
    
    # Create dict mapping names to their career info
    name_info = {}
    for _, row in careers_df.iterrows():
        clean_name = row['clean_name']
        if clean_name not in name_info:
            name_info[clean_name] = {
                'parties': set(),
                'start_dates': [],
                'full_names': set()
            }
        name_info[clean_name]['parties'].add(row['party'])
        name_info[clean_name]['start_dates'].append(row['start_date'])
        name_info[clean_name]['full_names'].add(row['full_name'])
    
    return careers_df, name_info

def calculate_seniority(speeches_df, careers_df, name_info):
    """Calculate seniority while verifying party matches"""
    # Clean names in speeches dataset
    speeches_df['clean_name'] = speeches_df['speaker_name'].apply(clean_name_for_matching)
    
    # Convert speech date to datetime
    speeches_df['date'] = pd.to_datetime(speeches_df['date'])
    
    # Initialize seniority column
    speeches_df['seniority_days'] = None
    
    # Track matches and mismatches for debugging
    matches = {}
    mismatches = {}
    
    for idx, row in speeches_df.iterrows():
        if row['clean_name'] is None:
            continue
            
        clean_name = row['clean_name']
        
        # Check if speaker exists in career data
        if clean_name not in name_info:
            mismatches[row['speaker_name']] = "Not found in career data"
            continue
            
        # Get career info
        info = name_info[clean_name]
        
        # Verify party match if party is provided
        if pd.notna(row['speaker_party']) and row['speaker_party'] not in info['parties']:
            mismatches[row['speaker_name']] = f"Party mismatch: {row['speaker_party']} vs {info['parties']}"
            continue
        
        # Calculate seniority from earliest start date
        start_date = min(info['start_dates'])
        seniority = (row['date'] - start_date).days
        speeches_df.at[idx, 'seniority_days'] = max(0, seniority)
        
        # Track successful match
        matches[row['speaker_name']] = f"Matched to {info['full_names']}"
    
    # Print matching report
    print("\nMatching Report:")
    print(f"Successfully matched: {len(matches)} speakers")
    print(f"Failed to match: {len(mismatches)} speakers")
    print("\nSample of successful matches:")
    for name, match_info in list(matches.items())[:10]:
        print(f"- {name}: {match_info}")
    print("\nFailed matches:")
    for name, reason in mismatches.items():
        print(f"- {name}: {reason}")
        
    return speeches_df
```

```python
# Load and process the data
careers_df, name_info = create_career_timeline(careers_df)
speeches_df = calculate_seniority(speeches_df, careers_df, name_info)

# Save results
output_path = os.path.join(os.path.dirname(speeches_path), 
                          'speeches_2014-2024_speakers_parties_fixed_with_seniority.csv')
speeches_df.to_csv(output_path, sep=';', index=False)

# Show statistics
print("\nSeniority Statistics:")
print(speeches_df[speeches_df['seniority_days'] >= 0]['seniority_days'].describe())
```
