```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import json
import ast
import os
import glob

# Function to check if a row contains an error
def check_error(result_str):
    try:
        result_dict = ast.literal_eval(result_str) if isinstance(result_str, str) else result_str
        if isinstance(result_dict, dict) and result_dict.get('status') == 'error':
            return True
        return False
    except:
        return True  # If we can't parse it, consider it an error

# Function to analyze a single CSV file
def analyze_csv(file_path):
    # Extract filename without extension for display
    filename = os.path.basename(file_path)
    print(f"\n{'='*80}\nAnalyzing: {filename}\n{'='*80}")
    
    try:
        # First try with semicolon delimiter
        df = pd.read_csv(file_path, sep=';')
        
        # Check if we have the expected columns
        required_cols = ['result', 'is_timeout', 'final_label', 'count']
        missing_cols = [col for col in required_cols if col not in df.columns]
        
        if missing_cols:
            # Try with comma delimiter if semicolon didn't work correctly
            df = pd.read_csv(file_path)
            missing_cols = [col for col in required_cols if col not in df.columns]
            
            if missing_cols:
                print(f"Error: Missing required columns {missing_cols}")
                return None
        
        # Find rows with errors or timeouts
        timeout_rows = df[df['is_timeout'] == 1]
        error_rows = df[df['result'].apply(check_error)]

        # Print excluded rows
        print("Rows excluded due to timeout:")
        if len(timeout_rows) > 0:
            print(timeout_rows[['index', 'file_id', 'speaker_name']].head())
            if len(timeout_rows) > 5:
                print(f"...and {len(timeout_rows) - 5} more")
        else:
            print("None")

        print("\nRows excluded due to errors:")
        if len(error_rows) > 0:
            print(error_rows[['index', 'file_id', 'speaker_name']].head())
            if len(error_rows) > 5:
                print(f"...and {len(error_rows) - 5} more")
        else:
            print("None")

        # Filter out rows where is_timeout is 1 or result contains an error
        df_filtered = df[(df['is_timeout'] != 1) & (~df['result'].apply(check_error))]

        # Print summary of excluded rows
        total_rows = len(df)
        excluded_rows = total_rows - len(df_filtered)
        print(f"\nTotal rows in dataset: {total_rows}")
        print(f"Total rows excluded: {excluded_rows} ({excluded_rows/total_rows:.2%})")
        print(f"Remaining rows for analysis: {len(df_filtered)}")

        # Extract the ground truth labels (1 for ad hominem, 0 for not ad hominem)
        y_true = df_filtered['final_label'].values

        # Extract the predicted labels (1 if LLM found ad hominem, 0 otherwise)
        y_pred = df_filtered['count'].apply(lambda x: 1 if x > 0 else 0).values

        # Calculate metrics
        accuracy = accuracy_score(y_true, y_pred) * 100
        precision = precision_score(y_true, y_pred, zero_division=0) * 100
        recall = recall_score(y_true, y_pred, zero_division=0) * 100
        f1 = f1_score(y_true, y_pred, zero_division=0) * 100

        # Print metrics
        print(f"\nAccuracy: {accuracy:.2f}%")
        print(f"Precision: {precision:.2f}%")
        print(f"Recall: {recall:.2f}%")
        print(f"F1 Score: {f1:.2f}%")

        # Generate and print confusion matrix
        cm = confusion_matrix(y_true, y_pred)
        print("\nConfusion Matrix:")
        print(f"True Negative: {cm[0][0]}")
        print(f"False Positive: {cm[0][1]}")
        print(f"False Negative: {cm[1][0]}")
        print(f"True Positive: {cm[1][1]}")
        
        # Visualize the confusion matrix
        plt.figure(figsize=(10, 7))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                    xticklabels=['No Ad Hominem', 'Ad Hominem'],
                    yticklabels=['No Ad Hominem', 'Ad Hominem'])
        plt.xlabel('Predicted')
        plt.ylabel('Actual')
        plt.title(f'Confusion Matrix for {filename}')
        plt.tight_layout()
        plt.show()
        
        # Return metrics for sorting later
        return {
            'filename': filename,
            'accuracy': accuracy,
            'precision': precision,
            'recall': recall,
            'f1': f1,
            'confusion_matrix': cm
        }
    except Exception as e:
        print(f"Error processing file {filename}: {e}")
        return None

# Main function to loop through all CSV files
def main():
    base_path = r'C:\Users\anouk\OneDrive - UvA\Documenten\GitHub\the good one\tweede-kamer\modal\results on labeled sample'
    csv_files = glob.glob(os.path.join(base_path, '**', '*.csv'), recursive=True)
    
    # File to exclude
    exclude_file = r'C:\Users\anouk\OneDrive - UvA\Documenten\GitHub\the good one\tweede-kamer\modal\results on labeled sample\Llama\English\results_tk_debate_NL_Llama4_zeroshot-prompt_EN_2025-05-15_05-54-35.csv'
    exclude_filename = os.path.basename(exclude_file)
    
    # Filter out the excluded file
    filtered_files = [f for f in csv_files if os.path.basename(f) != exclude_filename]
    
    print(f"Found {len(csv_files)} CSV files, analyzing {len(filtered_files)} files (excluding {exclude_filename})")
    
    results = []
    
    for file_path in filtered_files:
        result = analyze_csv(file_path)
        if result:
            results.append(result)
    
    # Add these lines to generate the tables
    print_sorted_table(results, 'f1', "SUMMARY OF RESULTS (SORTED BY F1 SCORE)")
    print_sorted_table(results, 'accuracy', "SORTED BY ACCURACY")
    print_sorted_table(results, 'precision', "SORTED BY PRECISION")
    print_sorted_table(results, 'recall', "SORTED BY RECALL")
    
# Function to print a table of results sorted by a specific metric
def print_sorted_table(results, metric, title):
    sorted_results = sorted(results, key=lambda x: x[metric], reverse=True)
    metric_name = {'f1': 'F1 Score', 'accuracy': 'Accuracy', 'precision': 'Precision', 'recall': 'Recall'}
    
    print(f"\n\n{title}")
    print("="*80)
    print(f"{'Filename':<60} {metric_name[metric]:<10}")
    print("-"*72)
    for r in sorted_results:
        print(f"{r['filename']:<60} {r[metric]:<10.2f}")

if __name__ == "__main__":
    main()
```

```python
def visualize_metrics(results):
    """
    Create bar charts for each metric comparing all models.
    """
    # Extract data for plotting
    filenames = [r['filename'] for r in results]
    # Create shorter labels by removing common prefixes and extensions
    short_labels = [f.replace('results_tk_debate_NL_', '').replace('.csv', '') for f in filenames]
    
    accuracies = [r['accuracy'] for r in results]
    precisions = [r['precision'] for r in results]
    recalls = [r['recall'] for r in results]
    f1_scores = [r['f1'] for r in results]
    
    # Create figure with multiple subplots
    fig, axes = plt.subplots(2, 2, figsize=(20, 16))
    fig.suptitle('Performance Metrics for Ad Hominem Detection Models', fontsize=16)
    
    # Accuracy plot
    sorted_indices = sorted(range(len(accuracies)), key=lambda i: accuracies[i], reverse=True)
    axes[0, 0].bar(range(len(accuracies)), [accuracies[i] for i in sorted_indices], color='skyblue')
    axes[0, 0].set_title('Accuracy (%)')
    axes[0, 0].set_xticks(range(len(accuracies)))
    axes[0, 0].set_xticklabels([short_labels[i] for i in sorted_indices], rotation=45, ha='right')
    axes[0, 0].set_ylim(0, 100)
    axes[0, 0].grid(axis='y', linestyle='--', alpha=0.7)
    # Add value labels on top of bars
    for i, v in enumerate([accuracies[i] for i in sorted_indices]):
        axes[0, 0].text(i, v + 1, f'{v:.1f}%', ha='center')
    
    # Precision plot
    sorted_indices = sorted(range(len(precisions)), key=lambda i: precisions[i], reverse=True)
    axes[0, 1].bar(range(len(precisions)), [precisions[i] for i in sorted_indices], color='lightgreen')
    axes[0, 1].set_title('Precision (%)')
    axes[0, 1].set_xticks(range(len(precisions)))
    axes[0, 1].set_xticklabels([short_labels[i] for i in sorted_indices], rotation=45, ha='right')
    axes[0, 1].set_ylim(0, 100)
    axes[0, 1].grid(axis='y', linestyle='--', alpha=0.7)
    for i, v in enumerate([precisions[i] for i in sorted_indices]):
        axes[0, 1].text(i, v + 1, f'{v:.1f}%', ha='center')
    
    # Recall plot
    sorted_indices = sorted(range(len(recalls)), key=lambda i: recalls[i], reverse=True)
    axes[1, 0].bar(range(len(recalls)), [recalls[i] for i in sorted_indices], color='salmon')
    axes[1, 0].set_title('Recall (%)')
    axes[1, 0].set_xticks(range(len(recalls)))
    axes[1, 0].set_xticklabels([short_labels[i] for i in sorted_indices], rotation=45, ha='right')
    axes[1, 0].set_ylim(0, 100)
    axes[1, 0].grid(axis='y', linestyle='--', alpha=0.7)
    for i, v in enumerate([recalls[i] for i in sorted_indices]):
        axes[1, 0].text(i, v + 1, f'{v:.1f}%', ha='center')
    
    # F1 Score plot
    sorted_indices = sorted(range(len(f1_scores)), key=lambda i: f1_scores[i], reverse=True)
    axes[1, 1].bar(range(len(f1_scores)), [f1_scores[i] for i in sorted_indices], color='mediumpurple')
    axes[1, 1].set_title('F1 Score (%)')
    axes[1, 1].set_xticks(range(len(f1_scores)))
    axes[1, 1].set_xticklabels([short_labels[i] for i in sorted_indices], rotation=45, ha='right')
    axes[1, 1].set_ylim(0, 100)
    axes[1, 1].grid(axis='y', linestyle='--', alpha=0.7)
    for i, v in enumerate([f1_scores[i] for i in sorted_indices]):
        axes[1, 1].text(i, v + 1, f'{v:.1f}%', ha='center')
    
    plt.tight_layout()
    plt.subplots_adjust(top=0.92)
    plt.show()
    
    # Also create a single plot comparing all metrics
    plt.figure(figsize=(14, 10))
    bar_width = 0.2
    x = np.arange(len(filenames))
    
    # Sort by F1 score for the combined chart
    sorted_indices = sorted(range(len(f1_scores)), key=lambda i: f1_scores[i], reverse=True)
    sorted_labels = [short_labels[i] for i in sorted_indices]
    
    plt.bar(x - 1.5*bar_width, [accuracies[i] for i in sorted_indices], bar_width, label='Accuracy', color='skyblue')
    plt.bar(x - 0.5*bar_width, [precisions[i] for i in sorted_indices], bar_width, label='Precision', color='lightgreen')
    plt.bar(x + 0.5*bar_width, [recalls[i] for i in sorted_indices], bar_width, label='Recall', color='salmon')
    plt.bar(x + 1.5*bar_width, [f1_scores[i] for i in sorted_indices], bar_width, label='F1 Score', color='mediumpurple')
    
    plt.xlabel('Model')
    plt.ylabel('Score (%)')
    plt.title('Comparison of All Metrics (Sorted by F1 Score)')
    plt.xticks(x, sorted_labels, rotation=45, ha='right')
    plt.legend()
    plt.ylim(0, 100)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

# Add this to your main function:
def main():
    # ...existing code...
    
    # After collecting results
    visualize_metrics(results)
    
    # Print tables sorted by each metric
    print_sorted_table(results, 'f1', "SUMMARY OF RESULTS (SORTED BY F1 SCORE)")
    print_sorted_table(results, 'accuracy', "SORTED BY ACCURACY")
    print_sorted_table(results, 'precision', "SORTED BY PRECISION")
    print_sorted_table(results, 'recall', "SORTED BY RECALL")
```

```python
def create_metric_plots(results):
    """
    Create individual bar plots for each metric (Accuracy, Precision, Recall, F1 Score).
    """
    # Extract data for plotting
    filenames = [r['filename'] for r in results]
    # Create shorter labels by removing common prefixes and extensions
    short_labels = [f.replace('results_tk_debate_NL_', '').replace('.csv', '') for f in filenames]
    
    # Define metrics to plot
    metrics = {
        'accuracy': 'Accuracy (%)',
        'precision': 'Precision (%)',
        'recall': 'Recall (%)',
        'f1': 'F1 Score (%)'
    }
    
    # Colors for each metric
    colors = {
        'accuracy': 'skyblue',
        'precision': 'lightgreen',
        'recall': 'salmon',
        'f1': 'mediumpurple'
    }
    
    # Create individual plots for each metric
    for metric_key, metric_title in metrics.items():
        plt.figure(figsize=(14, 8))
        
        # Get metric values
        values = [r[metric_key] for r in results]
        
        # Sort by the metric value
        sorted_indices = sorted(range(len(values)), key=lambda i: values[i], reverse=True)
        sorted_labels = [short_labels[i] for i in sorted_indices]
        sorted_values = [values[i] for i in sorted_indices]
        
        # Plot bars
        bars = plt.bar(range(len(sorted_values)), sorted_values, color=colors[metric_key])
        
        # Add value labels on top of bars
        for i, v in enumerate(sorted_values):
            plt.text(i, v + 1, f'{v:.1f}%', ha='center')
        
        # Set labels and title
        plt.title(f'{metric_title} for Ad Hominem Detection Models', fontsize=14)
        plt.ylabel(metric_title)
        plt.xticks(range(len(sorted_values)), sorted_labels, rotation=45, ha='right')
        plt.ylim(0, 100)  # Set y-axis from 0 to 100%
        plt.grid(axis='y', linestyle='--', alpha=0.7)
        plt.tight_layout()
        plt.show()

# Run the function - assuming 'results' contains your analysis results
create_metric_plots(results)
```
